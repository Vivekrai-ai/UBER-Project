```python
Files=[
{"file":"map_cities"},
{"file":"map_cancellation_reasons"},
{"file":"bulk_rides"},
{"file":"map_payment_methods"},
{"file":"map_ride_statuses"},
{"file":"map_vehicle_makes"},
{"file":"map_vehicle_types"}
]
 
```

```python
for file in Files:
    df= spark.read.format("json").option("inferschema","true").option("multiLine","true").load(f"abfss://uberbronze@vivostorageaccountname.dfs.core.windows.net/{file["file"]}.json")
                                                                                            
    df.write.format("delta").mode("append").saveAsTable(f"uber_project.bronze.{file["file"]}")
```
